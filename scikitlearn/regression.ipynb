```python
import yfinance as yf
import math

df = yf.download("GOOGL", start="2020-01-01", end="2020-12-31")
df = df[['Open', 'High', 'Low', 'Close', 'Volume']]
df['HL_PCT'] = (df['High'] - df['Close']) / df['Close'] * 100.0
df['PCT_change'] = (df['Close'] - df['Open']) / df['Open'] * 100.0
df = df[['HL_PCT', 'PCT_change', 'Volume', 'Close']]

#print(df.head())
forecast_col = 'Close'
df.fillna(-99999, inplace=True)  # Replace NaN with -99999
forecast_out = int(math.ceil(0.01 * len(df)))  # Forecast for 10%
df['label'] = df[forecast_col].shift(-forecast_out)
df.dropna(inplace=True)  # Drop rows with NaN values after shifting
df.tail() 
```
