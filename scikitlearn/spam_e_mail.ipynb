```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report
import requests
```

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00228/smsspamcollection.zip"
response = requests.get(url)

# Write the content to a zip file
with open("smsspamcollection.zip", "wb") as f:
    f.write(response.content)

# Extract the zip file
import zipfile
with zipfile.ZipFile("smsspamcollection.zip", 'r') as zip_ref:
    zip_ref.extractall(".")
```

```python
df = pd.read_csv("SMSSpamCollection", sep="\t", header=None, names=["label", "message"])
df['label'] = df["label"].map({"ham": 0, "spam": 1})
print("dataset sample: ",df.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(df["message"], df["label"], test_size=0.2, random_state=42)

```

```python
#vectorize
vectorizer = TfidfVectorizer(stop_words='english', max_features=5000)
X_train_vectorized = vectorizer.fit_transform(X_train)
X_test_vectorized = vectorizer.transform(X_test)
```

```python
#train model

model=MultinomialNB()
model.fit(X_train_vectorized, y_train)
```

```python
#evaluate
y_pred = model.predict(X_test_vectorized)
print("accuracy: ", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
#sample text

text_msg=[
    "hot milfs around your area waiting for you to have chat and sex with. click on the link below to chat.",
    "Hey, are we still meeting for lunch tomorrow?",
    "Congratulations! You’ve won a free iPhone, click the link to claim now!",
    "Bitch, send me the assignment file.",
    "here is the assignment, click the link to open"
]
```

```python
test_vecs = vectorizer.transform(text_msg)
predictions = model.predict(test_vecs)
print("\n custum predictions: ")
for msg, pred in zip(text_msg, predictions):
  print(f"Message: {msg}\nPrediction: {'SPAM' if pred == 1 else 'HAM'}\n")
```
