```python
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import  r2_score
from sklearn.preprocessing import PolynomialFeatures
from sklearn.ensemble import HistGradientBoostingRegressor,RandomForestRegressor
import joblib


housing=datasets.fetch_california_housing()


x=housing.data #features
y=housing.target #lable

print(x.shape)

poly = PolynomialFeatures()
x=poly.fit_transform(x) #transform the features to polynomial features

print(x.shape)

x_train, x_test, y_train, y_test = train_test_split(
     x,y,test_size=0.2 ,random_state=432) #split data into 20% training and 80% testing sets 


model = HistGradientBoostingRegressor(max_iter=350, learning_rate=0.05) #create the model
model.fit(x_train,y_train) #train the model

joblib.dump(model, 'model.joblib') #save the model
y_pred = model.predict(x_test) #predict the model
r2=r2_score(y_test, y_pred) #check the accuracy of the model

print(r2)
```
