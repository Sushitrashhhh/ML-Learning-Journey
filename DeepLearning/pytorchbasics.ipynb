## DAY1 W/ PYTORCH

```python
import torch, torch.nn as nn, numpy as np

#1. provide training data
# We need to format the data correctly for PyTorch.
# It expects tensors, and the shape needs an extra dimension.
x_train = torch.tensor([[-1.0], [0.0], [1.0], [2.0], [3.0], [4.0]], dtype=torch.float32)
y_train = torch.tensor([[-3.0], [-1.0], [1.0], [3.0], [5.0], [7.0]], dtype=torch.float32)

# 2. Define the Neural Network Model
# In PyTorch, models are typically defined as Python classes.
# nn.Linear(1, 1) means it takes 1 input feature and produces 1 output feature.
class LRModel(nn.Module):
    def __init__(self):
        super(LRModel, self).__init__()
        self.linear=nn.Linear(1,1) # 1 input, 1 output

    def forward(self,x):
        return self.linear(x)
    
model=LRModel()    

# 3. Define Loss Function and Optimizer
# 'loss_fn' measures how wrong the model is.
# 'optimizer' defines how the model should update itself to get better.
loss_fn=nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01) # Stochastic Gradient Descent

# 4. Train the model 
# Pytorch requires a explicit loop.
epochs=1000
print("start training... ")

for epoch in range(epochs):
    # forward pass
    y_pred=model(x_train)

    #calculate
    loss = loss_fn(y_pred, y_train)

    # backward pass
    loss.backward()

    # update weights
    optimizer.step()

    # zero the gradients for the next iterartion
    optimizer.zero_grad()

    if (epoch+1)%100==0:
        print(f'Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}')

print('finished training')

#5. Make a Prediction
# Let's test it on a new number. What is 'y' when x=10?
# The answer should be close to 2*10 - 1 = 19

new_x=torch.tensor([[10.0]])
predicted_y=model(new_x)

print(f"\nModel prediction for x=10: {predicted_y.item()}")
```
